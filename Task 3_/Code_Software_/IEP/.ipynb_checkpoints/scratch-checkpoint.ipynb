```python
def calculate_wait_time(rho):
    c = configs.CHARGER_PROT_NUM
    c_factorial= (math.factorial(c))
    max_charge_time = configs.MAX_CHARGING_TIME
    mu = 1/max_charge_time
    lamb = rho * c * mu
    r = lamb/mu
    term1 = (r**c)/(c_factorial*(1-rho))

    term2 = 0
    for n in range(c):
        term2 += ((r**n)/(math.factorial(n)))
    term3 = (term1)/(c*mu*(1-rho))
    p0 = 1/(term1 + term2)
    t_wait = (term3)*(p0)
    return t_wait
```

```python
def od_approach():
    solvable_od, solvable_od_demand, infessible_od, infessible_od_demand, graph, d_adj,cap = layer_lib.layer1()
    facilities = np.zeros(len(graph.vs), dtype=int)
    # Set the i-th element in PRESET_STATION to 1
    for i in configs.PRESET_STATION:
        facilities[i] = 1
    sorted_od = layer_lib.sort_od_to_hazard(solvable_od, solvable_od_demand, infessible_od, infessible_od_demand, graph)
    data = None
    pre_charger_usage = np.zeros(len(graph.vs), dtype=int)
    new_cap = cap
    for key,value in sorted_od.items():
        if value == []:
            continue
        else:
            if hasattr(data, 'flow_size'):
                new_cap = cap - (data.flow_size).to_numpy()
                # new_cap=np.where(new_cap < 0, 0, new_cap)
            # print(pd.DataFrame(new_cap))
            value_arr = np.array(value)
            od_pairs = value_arr[:, 0:3]
            demand_list = value_arr[:, 3]
            if data is not None:
                facilities = data.charging_demand_size.iloc[:, 0].to_numpy()
                pre_charger_usage = data.charging_demand_size.iloc[:, 1].to_numpy()
            print("start solve layer 3")
            data = layer_lib.od_layer3(od_pairs, demand_list, graph, d_adj, facilities,pre_charger_usage=pre_charger_usage, cap=new_cap, data=data)
    return data
```

```python
def save_average_wait_time(model, data, debug, layer):
    for i in model.nodes_set:
        for k in model.OD_set:
            print(i,k,model.e[i,k].value)
    lamb = np.zeros([model.nodes_size])
    mu = np.zeros([model.nodes_size])
    t_ch = np.zeros([model.nodes_size])
    rho = np.zeros([model.nodes_size])
    p0 = np.zeros([model.nodes_size])
    t_wait = np.zeros([model.nodes_size])
    term1 = np.zeros([model.nodes_size])
    term2 = np.zeros([model.nodes_size])
    term3 = np.zeros([model.nodes_size])
    c = configs.CHARGER_PROT_NUM
    for i in model.nodes_set:
        if layer == 0:
            lamb[i]=(sum(model.y[i, k]*model.demand_list[k] for k in model.OD_set))
        elif layer == 3 or layer == 5:
            lamb[i] = (sum(model.y[i, k].value * model.demand_list[k] for k in model.OD_set))
        else:
            lamb[i] = (sum(model.y[i, k].value * model.demand_list[k] for k in model.OD_set))
    for i in model.nodes_set:
        if lamb[i] > 0:
            t_ch[i] = sum((model.demand_list[k]*model.e[i,k].value*constants.RECHARGE_TIME)/lamb[i] for k in model.OD_set)
            if t_ch[i] > 0:
                mu[i] = 1/t_ch[i]
            else:
                mu[i] = None
        else:
            t_ch[i] = None
            mu[i] = None
    print("lamb:",lamb)
    print("t_ch:",t_ch)
    print("mu:",mu)
    r = lamb/mu
    rho = lamb/(c*mu)
    c_factorial= Decimal(math.factorial(c))
    for i in model.nodes_set:
        if not math.isnan(r[i]):
            term1[i] = Decimal(r[i]**c)/(c_factorial*Decimal(1-rho[i]))
            if term1[i] == Decimal('Infinity'):
                p0[i] = 0
            else:
                term2[i] = 0
                for n in range(c):
                    term2[i] += int(Decimal(r[i]**n)/Decimal(math.factorial(n)))
                if term2[i] == Decimal('Infinity'):
                    p0[i] = 0
                else:
                    p0[i] = 1/(term1[i] + term2[i])
            term3[i] = Decimal(term1[i])/Decimal(c*mu[i]*(1-rho[i]))
            t_wait[i] = Decimal(term3[i])*Decimal(p0[i])
    print("term1:",term1)
    print("term2:",term2)
    print("term3:",term3)
    print("p0:",p0)
    print("t_wait:",t_wait)
    return t_wait
```

```python
save_average_wait_time(data.model,data,debug=False,layer=3)
```
